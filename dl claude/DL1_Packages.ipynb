```python
# Practical 1: study of DL packages (TensorFlow, Keras, PyTorch)
import tensorflow as tf, keras, torch
from keras.datasets import mnist

print(f"TensorFlow {tf.__version__} | Keras {keras.__version__} | PyTorch {torch.__version__}")
print("TF GPU:", bool(tf.config.list_physical_devices('GPU')), "| Torch GPU:", torch.cuda.is_available())

(xtr, ytr), (xte, yte) = mnist.load_data()
print("MNIST:", xtr.shape, xte.shape)
```

```python
# Same maths in both frameworks: tensor product + automatic differentiation
a = [[1., 2.], [3., 4.]]
print(tf.matmul(tf.constant(a), tf.constant(a)).numpy())
print(torch.tensor(a) @ torch.tensor(a))

x = tf.Variable(3.)
with tf.GradientTape() as tape:          # TF records ops on a "tape"
    y = x ** 2
print("TF    dy/dx:", tape.gradient(y, x).numpy())

w = torch.tensor(3., requires_grad=True)  # PyTorch builds the graph on the fly
(w ** 2).backward()
print("Torch dy/dx:", w.grad.item())
```

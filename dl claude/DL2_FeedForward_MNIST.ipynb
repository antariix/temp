```python
# Practical 2: feed-forward NN (MLP) on MNIST with SGD
import pandas as pd, matplotlib.pyplot as plt
from keras import Sequential, layers
from keras.datasets import mnist
from keras.optimizers import SGD
from sklearn.metrics import classification_report

(X, y), (Xt, yt) = mnist.load_data()
X, Xt = X.reshape(-1, 784) / 255., Xt.reshape(-1, 784) / 255.   # flatten + scale to [0,1]
```

```python
# Architecture: 784 -> 128 -> 64 -> 10 ; Dropout is the generalisation technique
model = Sequential([layers.Input((784,)),
                    layers.Dense(128, activation='relu'), layers.Dropout(0.2),
                    layers.Dense(64, activation='relu'),
                    layers.Dense(10, activation='softmax')])
model.compile(optimizer=SGD(0.1), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
H = model.fit(X, y, validation_data=(Xt, yt), epochs=10, batch_size=128)
```

```python
# Evaluate + plot
print(classification_report(yt, model.predict(Xt).argmax(1)))
pd.DataFrame(H.history).plot(title="Training loss and accuracy", xlabel="Epoch", ylabel="Loss / Accuracy")
plt.show()
```

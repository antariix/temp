```python
# Practical 3: CNN image classifier (MNIST per manual; switch DATA to cifar10 for the CIFAR variant)
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from keras import Sequential, layers
from keras.datasets import mnist, cifar10

DATA = mnist                                   # <- or cifar10
(X, y), (Xt, yt) = DATA.load_data()
if X.ndim == 3: X, Xt = X[..., None], Xt[..., None]      # MNIST is greyscale: add channel axis
X, Xt, y, yt = X / 255., Xt / 255., y.ravel(), yt.ravel()
```

```python
# Architecture: [Conv -> Pool] x2 -> Conv -> Flatten -> Dense
model = Sequential([layers.Input(X.shape[1:]),
                    layers.Conv2D(32, 3, activation='relu'), layers.MaxPooling2D(),
                    layers.Conv2D(64, 3, activation='relu'), layers.MaxPooling2D(),
                    layers.Conv2D(64, 3, activation='relu'),
                    layers.Flatten(), layers.Dense(64, activation='relu'), layers.Dense(10, activation='softmax')])
model.summary()
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
H = model.fit(X, y, epochs=5, validation_data=(Xt, yt))
```

```python
# Evaluate, predict one random test image, plot curves
print("Test loss, accuracy:", model.evaluate(Xt, yt))
i = np.random.randint(len(Xt))
plt.imshow(Xt[i].squeeze(), cmap='gray'); plt.title(f"Predicted: {model.predict(Xt[i:i+1]).argmax()} | True: {yt[i]}"); plt.show()

h = pd.DataFrame(H.history); _, ax = plt.subplots(1, 2, figsize=(13, 4))
h.filter(like='loss').plot(ax=ax[0], xlabel='Epoch'); h.filter(like='acc').plot(ax=ax[1], xlabel='Epoch'); plt.show()
```
